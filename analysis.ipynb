```python
order_items["discount_pct"] = order_items["discount_pct"].fillna(0)
```

```python
order_items["discount_pct"].isna().sum()
```

```python
order_items["unit_price"] = order_items["unit_price"].fillna(
    order_items["product_id"].map(
        products.set_index("product_id")["selling_price"]
    )
)
```

```python
order_items["unit_price"].isna().sum()
```

```python
order_items["net_sales"] = (
    order_items["quantity"]
    * order_items["unit_price"]
    * (1 - order_items["discount_pct"] / 100)
).round(2)

order_items["gross_profit"] = (
    order_items["net_sales"]
    - (order_items["quantity"] * order_items["unit_cost"])
).round(2)
```

```python
order_items[["quantity", "unit_price", "discount_pct", "net_sales", "gross_profit"]].head()
```

```python
# Calculate the main business KPIs for Tamkeen Market

# Total revenue generated from all order items
total_revenue = order_items["net_sales"].sum()

# Total profit after product costs
total_profit = order_items["gross_profit"].sum()

# Number of unique orders
total_orders = orders["order_id"].nunique()

# Number of unique customers
total_customers = customers["customer_id"].nunique()

# Number of products in the catalog
total_products = products["product_id"].nunique()

# Display the KPIs
print(f"Revenue: SAR {total_revenue:,.2f}")
print(f"Profit: SAR {total_profit:,.2f}")
print(f"Orders: {total_orders:,}")
print(f"Customers: {total_customers:,}")
print(f"Products: {total_products:,}")
```

```python
# Connect orders with order items so we can analyze sales by city
sales = order_items.merge(
    orders[["order_id", "customer_id", "city", "order_datetime"]],
    on="order_id",
    how="left"
)

# Calculate revenue and profit by city
city_performance = (
    sales.groupby("city")
    .agg(
        revenue=("net_sales", "sum"),
        profit=("gross_profit", "sum"),
        orders=("order_id", "nunique")
    )
    .sort_values("revenue", ascending=False)
)

city_performance
```

```python
# Add product information to our sales data
sales_products = sales.merge(
    products[["product_id", "category", "brand"]],
    on="product_id",
    how="left"
)

# Calculate performance by product category
category_performance = (
    sales_products.groupby("category")
    .agg(
        revenue=("net_sales", "sum"),
        profit=("gross_profit", "sum"),
        orders=("order_id", "nunique"),
        units_sold=("quantity", "sum")
    )
    .sort_values("revenue", ascending=False)
)

category_performance
```

```python
# Extract the month from each order date
sales["month"] = sales["order_datetime"].dt.to_period("M")

# Calculate monthly revenue and profit
monthly_performance = (
    sales.groupby("month")
    .agg(
        revenue=("net_sales", "sum"),
        profit=("gross_profit", "sum"),
        orders=("order_id", "nunique")
    )
    .sort_index()
)

monthly_performance
```

```python
# Calculate customer-level performance
customer_performance = (
    sales.groupby("customer_id")
    .agg(
        revenue=("net_sales", "sum"),
        profit=("gross_profit", "sum"),
        orders=("order_id", "nunique")
    )
    .sort_values("revenue", ascending=False)
)

# Show the top 10 customers by revenue
customer_performance.head(10)
```

```python
# Add marketing channel to the sales dataset
sales = sales.merge(
    orders[["order_id", "marketing_channel"]],
    on="order_id",
    how="left"
)

# Revenue and profit by marketing channel
channel_performance = (
    sales.groupby("marketing_channel")
    .agg(
        revenue=("net_sales", "sum"),
        profit=("gross_profit", "sum"),
        orders=("order_id", "nunique")
    )
    .sort_values("revenue", ascending=False)
)

display(channel_performance)
```

```python
# Average order value
average_order_value = sales.groupby("order_id")["net_sales"].sum().mean()

# Overall profit margin
profit_margin = (total_profit / total_revenue) * 100

# Return rate
return_rate = (returns["order_id"].nunique() / total_orders) * 100

print(f"Average Order Value: SAR {average_order_value:,.2f}")
print(f"Profit Margin: {profit_margin:.2f}%")
print(f"Return Rate: {return_rate:.2f}%")
```

```python
import sqlite3

# Create the SQLite database
conn = sqlite3.connect("tamkeen_market.db")

# Load all cleaned tables into SQLite
customers.to_sql("customers", conn, if_exists="replace", index=False)
products.to_sql("products", conn, if_exists="replace", index=False)
orders.to_sql("orders", conn, if_exists="replace", index=False)
order_items.to_sql("order_items", conn, if_exists="replace", index=False)
deliveries.to_sql("deliveries", conn, if_exists="replace", index=False)
returns.to_sql("returns", conn, if_exists="replace", index=False)
warehouses.to_sql("warehouses", conn, if_exists="replace", index=False)
marketing_channels.to_sql("marketing_channels", conn, if_exists="replace", index=False)

print("SQLite database created successfully.")
```

```python
# Executive KPIs using SQL
kpis = pd.read_sql_query("""
SELECT
    ROUND(SUM(net_sales), 2) AS total_revenue,
    ROUND(SUM(gross_profit), 2) AS total_profit,
    COUNT(DISTINCT order_id) AS total_orders
FROM order_items;
""", conn)

kpis
```

```python
# Revenue, profit, and orders by city
city_sql = pd.read_sql_query("""
SELECT
    o.city,
    ROUND(SUM(oi.net_sales), 2) AS revenue,
    ROUND(SUM(oi.gross_profit), 2) AS profit,
    COUNT(DISTINCT o.order_id) AS orders
FROM orders o
JOIN order_items oi
    ON o.order_id = oi.order_id
GROUP BY o.city
ORDER BY revenue DESC;
""", conn)

city_sql
```

```python
# Monthly revenue and profit
monthly_sql = pd.read_sql_query("""
SELECT
    strftime('%Y-%m', o.order_datetime) AS month,
    ROUND(SUM(oi.net_sales), 2) AS revenue,
    ROUND(SUM(oi.gross_profit), 2) AS profit
FROM orders o
JOIN order_items oi
    ON o.order_id = oi.order_id
GROUP BY month
ORDER BY month;
""", conn)

# Category performance
category_sql = pd.read_sql_query("""
SELECT
    p.category,
    ROUND(SUM(oi.net_sales), 2) AS revenue,
    ROUND(SUM(oi.gross_profit), 2) AS profit,
    SUM(oi.quantity) AS units_sold
FROM order_items oi
JOIN products p
    ON oi.product_id = p.product_id
GROUP BY p.category
ORDER BY revenue DESC;
""", conn)

print("MONTHLY PERFORMANCE")
display(monthly_sql)

print("CATEGORY PERFORMANCE")
display(category_sql)
```

```python
# Customer performance
customer_sql = pd.read_sql_query("""
SELECT
    o.customer_id,
    COUNT(DISTINCT o.order_id) AS orders,
    ROUND(SUM(oi.net_sales), 2) AS revenue,
    ROUND(SUM(oi.gross_profit), 2) AS profit
FROM orders o
JOIN order_items oi
    ON o.order_id = oi.order_id
GROUP BY o.customer_id
ORDER BY revenue DESC;
""", conn)

customer_sql.head(10)
```

```python
# Return analysis
return_sql = pd.read_sql_query("""
SELECT
    return_reason,
    COUNT(*) AS returns,
    ROUND(SUM(refund_amount), 2) AS refunds
FROM returns
GROUP BY return_reason
ORDER BY returns DESC;
""", conn)

# Delivery performance
delivery_sql = pd.read_sql_query("""
SELECT
    delivery_status,
    COUNT(*) AS deliveries,
    ROUND(AVG(actual_days), 2) AS avg_delivery_days,
    ROUND(AVG(shipping_cost), 2) AS avg_shipping_cost
FROM deliveries
GROUP BY delivery_status
ORDER BY deliveries DESC;
""", conn)

print("RETURN ANALYSIS")
display(return_sql)

print("DELIVERY PERFORMANCE")
display(delivery_sql)
```
